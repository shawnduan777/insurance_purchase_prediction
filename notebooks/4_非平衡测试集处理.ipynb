# 非平衡测试集处理——随机剔除

对于测试集，阳性样本过少会使模型评价指标无法真实反映机器学习模型的预测效果，机器学习模型总可以通过预测阴性样本来获得较高的准确率，因此需要平衡测试集中两类样本的比例。由于smote过采样法会在数据中加入并不真实存在的合成样本，会给数据集产生噪声，掩盖变量间的真实联系，因此对于测试集数据，本研究采用随机剔除一定数量的阴性样本来平衡两类样本数量。

划分后的测试集共有2645个样本，其中阴性样本2454个，阳性样本191个，样本严重不平衡。对于任何机器学习模型，其总可以通过将所有样本预测为阴性来获得92.7%的正确率，因此需要随机删除一定数量的阴性样本以缓解该问题。又考虑到在目前真实情况下，参与商业养老保险的群体仍旧占少数，因此测试集中阳性样本的数量占比不能过多，以此模拟实际应用中该模型对于真阳性样本的识别。综合考虑上述两种因素，本研究采用将测试集中阴性样本随机删除，直到剩余样本占测试集总样本的70%，阳性样本占测试集样本总数的30%。最终测试集中有阳性样本191个，阴性样本445个。

```python
import pandas as pd
```

### 8:2平衡测试集

```python
df = pd.read_excel("两年测试集.xlsx")

# 计算当前 0 和 1 的数量
count_0 = df[df['com_endow'] == 0].shape[0]
count_1 = df[df['com_endow'] == 1].shape[0]

# 计算在 80% 的情况下，0 类样本应该有多少
desired_count_0 = int(count_1 / 0.2) - count_1  # 0.2 是 1 样本的占比

# 保持 1 类样本不变，随机下采样 0 类样本
df_class_0 = df[df['com_endow'] == 0].sample(n=desired_count_0, random_state=42)
df_class_1 = df[df['com_endow'] == 1]

# 合并两类样本，形成新的数据框
df_balanced = pd.concat([df_class_0, df_class_1])

# 检查新的样本比例
print("新数据集中 0 的比例: {:.2f}".format(df_balanced['com_endow'].value_counts(normalize=True).get(0, 0)))
print("新数据集中 1 的比例: {:.2f}".format(df_balanced['com_endow'].value_counts(normalize=True).get(1, 0)))

df_balanced.to_excel("82平衡测试集.xlsx", index = False) 
```
