# 偏依赖图分析

偏依赖图分析代码见7_2:"XGBoost模型"

偏依赖图通过固定某一特征为特定值，让其他特征取所有可能的值进行组合，计算模型的平均输出。这种方法可以最大程度上消除其他特征对模型输出的影响，反映单一特征对模型预测的平均影响。本研究选取对7个购买商业养老保险行为具有重要影响的多值/连续变量，通过计算目标特征在不同水平下的平均效应，得到如下部分依赖曲线。

## 受教育程度

<div align="center">
    <img src="figures/偏依赖图edu.png" width="450">
</div>

受教育程度的shap值分布非常分散，表明受教育程度对商业养老保险购买行为有非常强大的预测能力。偏依赖图反映出受教育程度与商业养老保险的购买意愿有显著的正相关关系，当个体具有高中及以上学历时，其购买商业养老保险的概率会显著提高，而当学历为本科及以上时，学历的进一步提升对购买行为的影响将不再显著。低学历个体普遍收入水平及认识水平较低，没有经济能力去购买商业养老保险，也往往对商业养老保险没有清楚的认知，没有明确的养老规划。

## 消极情绪



<div align="center">
    <img src="figures/偏依赖图depression.png" width="450">
</div>

研究显示，情绪评分较低（经常感到沮丧等消极情绪）的个体购买商业养老保险的意愿明显低于情绪评分较高（很少感到沮丧等消极情绪）的个体。这可能由于情绪与个体财务、生活状况紧密相关，很少感到消极情绪的个体一般拥有较好的生活质量和稳健的财务情况，他们可能更有能力，也更有意愿投资与未来的生活。相反，经常感到沮丧的个体往往面临比较大的生活和财务压力，使其无暇估计对未来的养老规划。从消费者心理学的角度来看，很少感到沮丧的个体通常具备更大的心理弹性，这使得他们在从事理财活动和未来规划时更加理性和长远。而经常感到沮丧的个体收到情绪影响，普遍更倾向于短期思考，缺乏进行长远规划的能力。

## 房产数量

<div align="center">
    <img src="figures/偏依赖图house.png" width="450">
</div>

房产数量。当家庭拥有的房产数量在2套及以上时，个体购买商业养老保险的概率显著高于拥有0套或1套房产的个体。房产数量直接与家庭资产储备和财务能力相关，拥有两套及以上房产的家庭更有经济余力来购买商业养老保险来进一步保障其未来的生活水平，也更有财产规划与投资的需求。值得注意的是，家庭房产数量为1的样本购买商业养老保险的概率最低，甚至低于0套房产的样本。这可能是由于拥有一套房产的个体其住房主要用于自住，且与其财务状况紧密相连，这类群体更有可能将大部分资金用在购买住房上，且更有可能背负房贷等债务，没有能力与精力购买商业养老保险。而对于0套房产的个体，相较于1套房产的个体拥有更灵活的资金使用能力，更有可能关注到未来的不确定性，进而选择购买商业养老保险。

根据shap汇总图及偏依赖图，新型媒体与传统媒体使用情况对个体购买商业养老保险行为的影响强度与影响机制十分接近，新型媒体影响能力略强。两类媒体使用频率越高的个体，购买商业养老保险的概率越高。其机制可能为在使用新型媒体、传统媒体的过程中，个体有更大的概率接触到有关养老问题或商业养老保险的信息，这会对个体的养老意识和对商业养老保险的了解产生积极影响。由于新型媒体具有更好的交互性，个体可以在使用新型媒体的过程中通过与好友聊天，发表评论来获取更加贴近自身的信息，因此这类信息渠道在叠加了伙伴群体效应和社会成数效应后能够产生更大的影响能力。

## 女儿数量

<div align="center">
    <img src="figures/偏依赖图daughter.png" width="450">
</div>

女儿数量与商业养老保险的购买意愿呈现出复杂的非线性关系，这背后可能受到多种因素的共同影响。当个体拥有大于等于两个女儿时，其购买商业养老保险的意愿会显著降低。拥有多个女儿的家庭面临更大的经济支出和责任，例如女儿的教育、婚嫁费用等，这会挤占大部分资金，导致可运用到商业养老保险的资金减少。同时，受到传统家庭养老观念的影响，子女应承担部分甚至全部的养老责任，拥有更多女儿的个体可能更依赖于依靠子女养老，而不是寻求商业养老保险提供养老保障。拥有0个或1个女儿的个体抚育自子女的经济压力较小，拥有更充裕的资金用于保险消费。而拥有一个女儿的个体购买商业养老保险的概率最大，这可能由于女儿提供的代际情感支持与生活照料使得父母处于相对良好的心理和生理状况，他们拥有更积极的生活态度和对未来的乐观预期，因此更倾向于购买商业养老保险。

对于图中所示的在女儿数量足够多（大于5个）后购买概率急剧上升，经后续检查认为其受到极端值影响。在2645个测试集样本中，仅有5个样本女儿数量大于等于5个，因此本研究认为偏依赖图在女儿数量在大于5后出现的急剧上升不具备现实意义。（见代码文件5_最终训练集与测试集描述性统计）

## 年龄



<div align="center">
    <img src="figures/偏依赖图age.png" width="450">
</div>

年龄与商业养老保险的购买行为也呈现出较为复杂的影响关系，在年龄为18-50岁之间时，年龄变化对商业养老保险的购买概率几乎没有影响，处于50-55岁的个体购买商业养老保险的概率有一定提高，这可能是由于我国大部分商业养老保险的最高购买年龄在55-65岁之间，此类限制促使许多人在超过年龄上限前选择购买商业养老保险。而55岁和65岁时偏依赖图的两次断崖式下降，也反映了保险条款对年龄上限规定的影响。
