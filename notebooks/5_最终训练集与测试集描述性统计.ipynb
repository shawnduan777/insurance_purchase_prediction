# 最终数据集描述性统计

在进行模型训练前，对最终数据进行描述性统计，检查数据质量。

```python
import pandas as pd
import numpy as np

df_train = pd.read_excel("两年训练集过采样.xlsx")
means = df_train.mean()
stds = df_train.std()
mins = df_train.min()
maxs = df_train.max()
na_counts = df_train.isnull().sum()

stats_df_train = pd.DataFrame({
    'Mean' : means,
    'Std' : stds,
    'Min' : mins,
    'Max' : maxs,
    'na_count' : na_counts
    })
print(stats_df_train)

stats_df_train.to_excel("过采样两年训练集描述性统计.xlsx", index = True)  

df_test = pd.read_excel("73平衡测试集.xlsx")
means = df_test.mean()
stds = df_test.std()
mins = df_test.min()
maxs = df_test.max()
na_counts = df_test.isnull().sum()

stats_df_test = pd.DataFrame({
    'Mean' : means,
    'Std' : stds,
    'Min' : mins,
    'Max' : maxs,
    'na_count' : na_counts
    })
print(stats_df_test)

stats_df_test.to_excel("73平衡测试集描述性统计.xlsx", index = True) 
```
