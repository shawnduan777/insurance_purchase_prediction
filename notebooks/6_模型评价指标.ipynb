# 模型评价指标

对于二分类机器学习模型，常见的评价指标有混淆矩阵、精确率、召回率、ROC曲线及AUC值等。

## 混淆矩阵

根据样本的实际值和模型预测值，可以将最后的预测结果分为四类：TP（真阳性）、FP（假阳性）、TN（真阴性）、FN（假阴性）。将其用矩阵形式表达，即为混淆矩阵。

| |实际为阳性|实际为阴性|
|---|---|---|
|预测为阳性|TP|FP|
|预测为阴性|FN|TN|

### 精确率

精确率反映预测为阳性的样本中，实际为阳性的比例，计算公式为：

$$
Precision = \frac{TP}{TP+FP}
$$

在本研究中，精确率反映的是在被预测为会购买商业养老保险的群体中，真正会购买商业养老保险的群体的比例。

### 召回率

召回率反映所有实际为阳性的样本中，被成功识别为阳性样本的比例，计算公式为：

$$
Recall = \frac{TP}{TP+FN}
$$

在本研究中，召回率反映所有商业养老保险的潜在购买者中，有多大比例被成功识别为有购买意愿者。

### F1分数

F1分数是精确率和召回率的调和平均数，具体公式为：

$$
F1 = \frac{2\times (Precision\times Recall)}{Precision + Recall}
$$

F1分数是对模型整体上对阳性样本预测能力的综合反映，数值越接近1表示模型在这两方面的综合表现越好。

### ROC曲线及AUC

ROC曲线也是评价二分类模型性能的工具之一，其通过展示不同分类阈值下模型的真阳性率（召回率）和假阳性率的关系，反映模型整体预测性能。

假阳性率（FPR）计算公式为：

$$
FPR = \frac{FP}{FP+TN}
$$

真阳性率（TPR）计算公式为：

$$
TPR = \frac{TP}{TP+FN}
$$

对于一个二分类模型，模型会对测试集每个样本计算其属于阳性样本的概率。对于给定的分类阈值，模型会将概率大于阈值的样本预测为阳性，概率小于阈值的样本预测为阴性，由此可以计算出该阈值下的假阳性率与真阳性率。通过在0到1之间连续调整分类阈值，可以计算出一系列假阳性率与真阳性率。将假阳性率为横轴，真阳性率为纵轴，将所有数值对应的点连线得到的即为ROC曲线。将ROC图中的（0，0）到（1，1）的对角线相连，得到的线段反映随机猜测的表现，如果ROC曲线位于随机猜测线的上方，表示模型具有区分能力，曲线越靠近左上角表示区分能力越强。如果ROC曲线位于随机猜测线下方，表示模型的区分能力不如随机猜测的表现。

AUC值通常与ROC曲线结合使用，用以定量反映模型的整体预测性能。其定义为ROC曲线下方的面积，取值范围在0到1之间，AUC值越接近1，表示模型性能越好。
