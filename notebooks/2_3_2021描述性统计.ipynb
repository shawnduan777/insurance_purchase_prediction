第一步：导入2021清洗后数据（from 1_2）

```python
import pandas as pd
import numpy as np

df_2021 = pd.read_excel("2021清洗后数据.xlsx", index_col='id')
print(df_2021.head())
```

第二步：描述性统计与缺失值处理

```python
means = df_2021.mean()
stds = df_2021.std()
mins = df_2021.min()
maxs = df_2021.max()
na_counts = df_2021.isnull().sum()

stats_df_2021 = pd.DataFrame({
    'Mean' : means,
    'Std' : stds,
    'Min' : mins,
    'Max' : maxs,
    'na_count' : na_counts
    })
print(stats_df_2021)
```

第三步：导出数据

```python
stats_df_2021.to_excel("2021描述性统计.xlsx", index = True) 
```
