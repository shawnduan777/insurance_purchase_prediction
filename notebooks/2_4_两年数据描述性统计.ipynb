第一步：导入2021清洗后数据(from 1_2)、2023清洗后数据(from 1_1)并合并

```python
import pandas as pd
import numpy as np

df_2021 = pd.read_excel("2021清洗后数据.xlsx", index_col = 'id')
df_2023 = pd.read_excel("2023清洗后数据.xlsx", index_col = 'id')
df_2years = pd.concat([df_2023, df_2021], axis=0, ignore_index=True)
```

```python
第二步：两年数据描述性统计
```

```python
means = df_2years.mean()
stds = df_2years.std()
mins = df_2years.min()
maxs = df_2years.max()
na_counts = df_2years.isnull().sum()

stats_df_2years = pd.DataFrame({
    'Mean' : means,
    'Std' : stds,
    'Min' : mins,
    'Max' : maxs,
    'na_count' : na_counts
    })
print(stats_df_2years)
```

第三步：导出数据

```python
stats_df_2years.to_excel("两年数据描述性统计.xlsx", index = True) 
df_2years.to_excel("两年清洗后数据.xlsx", index = True) 
```
