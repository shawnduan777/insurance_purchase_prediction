# 模型调优与预测结果

对于本项目所研究的问题，由于阴、阳性样本处于高度不平衡，传统的评价指标正确率并不能很好的反映出模型的预测性能。考虑到在保险销售的实际应用中，更值得关注的是有意愿购买保险的个体，即阳性样本，因此对模型表现的评价应该侧重于在识别阳性样本的表现上。又因为对于保险销售来说，遗漏掉潜在客户的代价大于将保险推销给没有购买保险意愿的个体的代价，因此本研究将重点关注模型的阳性样本召回率表现，同时关注F1分数的表现情况，综合评价模型对阳性样本预测能力。在超参数调优中，所有模型均采用5折交叉验证以保证结果的稳健性。

#### 模型代码及训练细节见7_1、7_2、7_3

### LightGBM模型

| |精确率|召回率|F1得分|支持度|
|---|---|---|---|---|
|0|0.81|0.77|0.79|445|

LightGBM模型AUC值为0.7514，在对阳性样本的识别中，所有被预测为阳性的样本中有52%的样本为真阳性样本；在所有真阳性样本中，有58%的样本被正确识别，F1得分为0.55。

#### 在以上三种机器学习模型的预测结果中，XGBoost模型拥有最高的召回率（XGBoost：0.72、随机森林：0.55、LightGBM：0.58），表示在所有商业养老保险的潜在客户中，该模型成功识别了其中的72%，其他模型的该比例显著低于XGBoost模型表现，而这在实际保险销售中非常重要。虽然XGBoost模型的精确率在三个模中最低（XGBoost：0.46、随机森林：0.52、LightGBM：0.52），但与随机森林、LightGBM模型的表现差距不大，表示在所有被预测为会购买商业养老保险的样本中，有46%的样本实际会购买。相比之下，拥有最较精确率的LightGBM模型的召回率只有0.58，意味着该模型会遗漏42%的潜在客户，这在实际应用中会为保险销售方造成巨大的收入损失。综合来看，XGBoost模型拥有最高的F1得分（0.56）和AUC值（0.76）。因此，本文选取XGBoost模型为最优模型进行进一步分析。
