第一步：导入清洗后的2023数据（from 1_1）

```python
import pandas as pd
import numpy as np

df_2023 = pd.read_excel("2023清洗后数据.xlsx", index_col='id')
print(df_2023.head())
```

第二步：描述性统计与缺失值处理

```python
means = df_2023.mean()
stds = df_2023.std()
mins = df_2023.min()
maxs = df_2023.max()
na_counts = df_2023.isnull().sum()

stats_df_2023 = pd.DataFrame({
    'Mean' : means,
    'Std' : stds,
    'Min' : mins,
    'Max' : maxs,
    'na_count' : na_counts
    })
print(stats_df_2023)
```

第三步：导出数据

```python
stats_df_2023.to_excel("2023描述性统计.xlsx", index = True) 
```
