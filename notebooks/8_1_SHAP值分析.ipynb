# SHAP值分析

SHAP值分析代码见7_1：“XGBoost模型”

由于传统的机器学习模型普遍具有“黑盒”问题，即使得到较好的预测结果，也无法理解与解释其背后的运算过程。而在保险营销实务中，销售者不仅需要机器学习模型直接给出的预测结果，更需要理解背后哪些因素具有显著的预测能力，才能更好的辅助推销活动以及有针对性的优化模型及数据。因此本研究通过SHAP来对影响因素重要性进行排序，SHAP使用博弈论里的Shapley值计算特征对模型输出的边际贡献，来对机器学习的“黑盒模型”进行解释。

由SHAP汇总图可以看出，对居民是否购买商业养老保险具有重要预测能力的前十个特征为：受教育程度、是否居住在城市、是否具有基本医疗保险、是否具备金融素养、沮丧程度评分、家庭房产数量、新型媒体使用频率、女儿数量、传统媒体使用频率、年龄。对于二值变量，shap图可以较为清楚的看到变量取值对购买商业养老保险行为的影响，而对于多值或连续型变量，还需要结合偏依赖图来分析其背后的影响机制。因此8_1将重点分析“是否居住在城市”“是否具有基本养老保险”“是否具备金融素养”三个变量，其余变量将在8_2具体分析。

## 是否居住在城市

SHAP图显示，是否居住于城市的特征具有显著的区分度，取值为0和取值为1的样本shap值分界明显。表明居住在城市的个体更有可能购买商业养老保险。这可能是多种因素共同作用的结果：城市居民的收入水平往往更高，这让该群体更容易负担商业养老保险的费用；城市居民普遍有更高的金融素养与受教育水平，使得他们更容易接触到保险这类产品，也往往对商业养老保险有更清晰的认知，这种意识可能使城市居民将商业养老保险作为未来养老规划的一部分。从生活成本与稳定性来看，城市居民有更高的生活成本以及工作和收入的稳定性不如村镇居民，因此城市居民可能更担心未来的不确定性会影响自己的养老生活，所以购买商业养老保险以弥补未来可能出现的资金缺口。

## 是否具有基本养老保险

整体来看，具有基本养老保险的个体购买商业养老保险的意愿显著大于没有基本养老保险的个体。其中没有基本养老保险的样本的shap值分布非常分散且取值的绝对值明显大于具有基本养老保险样本。这表明该变量对非商业养老保险的潜在客户具有非常强大的排除能力。这可能是由于有基本养老保险的个体普遍比没有基本养老保险的个体具备相对充分的保险知识，更能准确认识到养老保险的作用，并且可能这类群体在生活中已经从基本养老保险中获益，因此更加信任商业养老保险。同时，受到特定的政策环境影响，有基本养老保险的个体通常为在职或曾在职的员工，并且加入了正规的社保体系，而没有基本养老保险的个体往往是没有固定工作、灵活就业者或者失业待业者。这导致有基本养老保险的个体普遍拥有更稳定的收入来源和一定的经济基础，使他们更有能力选择商业养老保险作为养老补充。

## 是否具备金融素养

SHAP图显示，具备金融素养的个体更有可能购买商业养老保险。由于商业养老保险本就属于金融产品的一种，因此具备金融素养这一特征对购买商业养老保险的行为具有直接影响。同时由于本文对该特征的构造来源于受访者是否参与任何一种风险投资，因此对该问题做出肯定回答的个体可能具有更强的风险偏好以及经济能力，而商业养老保险本就具有一定的投资属性，可能会有一部分个体将商业养老保险作为整体投资组合的一部分进行风险分散。
