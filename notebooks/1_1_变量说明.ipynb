# 变量说明

#### 变量构造过程见文件1_2，1_3。

## 总体说明



本项目共选取1个被解释变量与22个解释变量。解释变量为：年龄、婚姻状况、家庭规模、受教育情况、儿子数量、女儿数量、房产数量、代际经济支持、代际情感支持、代际生活照料、传统媒体使用频率、新型媒体使用频率、社会互动频率、居住地类型、家庭年总收入、幸福感、负面情绪、自评健康、是否有基本养老保险、是否有基本医疗保险、养老观念、金融素养。被解释变量为：是否购买了商业养老保险。

## 代际支持、传统媒体使用频率、新型媒体使用频率

其中，由于只有2023年的问卷有关于代际支持的问题，并且对该问题的回答存在严重缺失值，因此将三个代际支持变量从本文中剔除。对于传统媒体的使用频率，问卷中有关与受访者报纸、杂志、广播、电视使用频率的问题，评分分为1、2、3、4、5，1为从不，5为非常频繁。将每个样本关于这四个问题回答的评分取算术平均数作为本研究传统媒体使用频率变量的取值。对于新型媒体使用频率，2023年问卷中有关于此问题的直接回答，评分标准与传统媒体使用情况一致，2021年问卷中有关于受访者互联网、手机定制消息使用频率的问题，评分标准与传统媒体使用情况一致，本研究依然采用取算数平均数的方法构造新型媒体使用频率变量，2023年问卷将此二问题合二为一，可以直接使用。

## 养老观念

对于养老观念变量，问卷中有“你认为父母的养老问题主要应由谁负责？”的问题，其中回答“主要由政府负责”赋值为1，回答“主要由子女负责”赋值为2，回答“主要由老人自己负责”赋值为3，回答“由三方责任均摊”赋值为4。但由于离散的数值之间本身具有顺序关系，但对该问题的四个回答并没有明确的顺序逻辑，因此如果不加处理直接应用可能会降低机器学习模型的预测效果。因此本研究对该变量采用了独热编码处理，将对该问题的回答分解为4个变量，每个变量仅反映受访者的回答是否为该类型。同时，由于本研究所采用的三类机器学习模型其本质均为决策树模型，区别于传统的线性模型，其对于二元分类变量的处理仅根据分裂点将数据划分为不同子集，不会强行假设两者之间的影响是一个线性差值。因此对于其他二分类变量，本文未使用独热编码处理，确保了较低的特征维度，且不会因独热编码产生稀疏性问题。

## 金融素养

对于金融素养变量，本研究根据调查问卷中有关“您家目前是否从事下列投资活动？-没有任何投资活动/股票/基金/债券/期货/权证/炒房/外汇投资/其他”问题的回答进行构造，如果受访者从事上述任何一项投资活动，则将其标记为1（具备金融素养），否则为0。

## 全部变量基本信息

| 字段名称 | 变量名称 | 详细信息 | 变量编码 |
|---|---|---|---|
| `com_endow` | 商业养老保险购买情况 | 是否购买商业养老保险 | 购买 = 1；未购买 = 0 |
| `age` | 年龄 | 以 2026 年为基准计算的受访者年龄 | 年龄（岁） |
| `marriage` | 婚姻状况 | 受访者的婚姻状况 | 已婚 = 1；未婚或离婚 = 0 |
| `member` | 家庭规模 | 家庭成员数量 | 家庭成员人数 |
| `edu` | 受教育程度 | 受访者达到的最高教育水平 | 义务教育及以下 = 1；高中及专科 = 2；本科 = 3；研究生及以上 = 4 |
| `son` | 儿子数量 | 包括亲生、非亲生及已过世儿子的数量 | 儿子人数 |
| `daughter` | 女儿数量 | 包括亲生、非亲生及已过世女儿的数量 | 女儿人数 |
| `house` | 房产数量 | 与受访者共同居住的家庭成员拥有的房产总数 | 房产数量 |
| `media_old` | 传统媒体使用频率 | 受访者使用传统媒体的频率 | 从不 = 1；很少 = 2；有时 = 3；经常 = 4；非常频繁 = 5 |
| `media_new` | 新型媒体使用频率 | 受访者使用新型媒体的频率 | 从不 = 1；很少 = 2；有时 = 3；经常 = 4；非常频繁 = 5 |
| `social` | 社会互动频率 | 受访者参与社会互动的频率 | 从来不 = 1；一年一次或更少 = 2；一年几次 = 3；约一个月一次 = 4；一个月几次 = 5；一周 1–2 次 = 6；几乎每天 = 7 |
| `urban` | 居住地类型 | 受访者居住地属于城镇还是乡村 | 城镇 = 1；乡村 = 0 |
| `family_income` | 家庭年总收入 | 与受访者共同居住的家庭成员年总收入 | 年总收入；超过 100 万元的统一编码为 999996 |
| `happy` | 幸福感 | 受访者对自身幸福程度的评价 | 非常不幸福 = 1；比较不幸福 = 2；说不上不幸福 = 3；比较幸福 = 4；非常幸福 = 5 |
| `depression` | 负面情绪频率 | 受访者感受到负面情绪的频率 | 总是 = 1；经常 = 2；有时 = 3；很少 = 4；从不 = 5 |
| `self_health` | 自评健康状况 | 受访者对自身健康状况的评价 | 很不健康 = 1；比较不健康 = 2；一般 = 3；比较健康 = 4；非常健康 = 5 |
| `basic_endow` | 基本养老保险参保情况 | 受访者是否参加基本养老保险 | 已参加 = 1；未参加 = 0 |
| `basic_medicine` | 基本医疗保险参保情况 | 受访者是否参加基本医疗保险 | 已参加 = 1；未参加 = 0 |
| `opinion_gov` | 政府养老责任认知 | 受访者是否认为养老问题应由政府负责 | 是 = 1；否 = 0 |
| `opinion_kid` | 子女养老责任认知 | 受访者是否认为养老问题应由子女负责 | 是 = 1；否 = 0 |
| `opinion_self` | 个人养老责任认知 | 受访者是否认为养老问题应由老人自己负责 | 是 = 1；否 = 0 |
| `opinion_all` | 共同养老责任认知 | 受访者是否认为养老问题应由政府、子女和老人共同负责 | 是 = 1；否 = 0 |
| `investment` | 金融素养 | 受访者是否具有金融素养 | 具有 = 1；不具有 = 0 |
